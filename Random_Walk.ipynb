## 1.Generar y graficar 1 caminata aleatoria

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

n_steps =200
steps = np.random.choice([-1,1], size=n_steps)
price =100 + np.cumsum(steps)

plt.figure(figsize=(10,5))
plt.plot(price)
plt.title("Random Walk de un precio")
plt.xlabel("Tiempo")
plt.ylabel("Precio")
plt.show()

```

## 2. Generar y graficar la distribución final de N caminatas aleatorias

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)

n_sim =50000
n_steps =200

final_positions = []

for _ in range(n_sim):
    steps = np.random.choice([-1,1], size=n_steps)
    final_positions.append(np.sum(steps))

plt.hist(final_positions, bins=80)
plt.title("Distribución de posiciones finales")
plt.show()

```

## 3. Calcular la varianza de los resultados para diferentes largos de caminatas

```python
import numpy as np

np.random.seed(0)

n_sim =2000
times = [10,50,100,200,400]

for t in times:
    finals = []
    for _ in range(n_sim):
        steps = np.random.choice([-1,1], size=t)
        finals.append(np.sum(steps))
print(f"t={t}, var≈{np.var(finals):.2f}")

```

## 4.Generar y graficar una aproximación de un Mov. Browniano con un random walk 

```python
np.random.seed(42)

T =1
n =1000
dt = T/n

increments = np.random.normal(0, np.sqrt(dt), size=n)
W = np.cumsum(increments)

plt.plot(W)
plt.title("Aproximación a movimiento browniano")
plt.show()

```

## 5. Graficar caminatas aleatorias

```python
plt.figure(figsize=(10,6))

paths=[]

for _ in range(200):
    steps = np.random.choice([-1,1], size=200)
    path = np.cumsum(steps)
    paths.append(path)
    plt.plot(path, alpha=0.2)

paths = np.mean(paths,axis=0)

plt.plot(paths, linewidth=2, color='black')
plt.title("200 Random Walks")
plt.show()
```
