# Introducción a Montecarlo en Python

```python
import numpy as np

import matplotlib.pyplot as plt
```

## 1. Calculadora de integrales mediante Monte Carlo


```python
def funcion_a_integral(x):
    return x**2

a=2
b=5

N = 100000

x=np.random.uniform(a,b,N)
y=np.random.uniform(0,funcion_a_integral(b),N)
n_bajo_curva = np.sum(y<funcion_a_integral(x))
respuesta = (b-a) * funcion_a_integral(b) * n_bajo_curva / N
respuesta
```

Generamos de manera aleatoria múltiples puntos en un rango determinado, luego vimos cuántos de ellos estaban por debajo de la función y de esa manera simulamos con Montecarlo el cálculo de una integral.

```python
plt.plot(x,y,'.')
plt.plot(np.linspace(a,b,100),funcion_a_integral(np.linspace(a,b,100)),'r')
plt.show()
```

## 2. Estimación de π usando el método del círculo


```python
# Ahora hacemos lo mismo pero para calcular el valor de pi
def funcion_circulo(x):
    return np.sqrt(1- x**2)

a=-1
b=1
N=1000000
x=np.random.uniform(a,b,N)
y=np.random.uniform(0,1,N)

n_bajo_curva = np.sum(y<funcion_circulo(x))
respuesta = 4 * n_bajo_curva / N
respuesta
```

```python
plt.plot(x,y,'.')
plt.plot(np.linspace(a,b,100),funcion_circulo(np.linspace(a,b,100)),'r')
plt.show()
```
